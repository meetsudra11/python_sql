```python
from itertools import permutations

def clock():
    digits = []

    for i in range(4):
        digits.append(int(input("enter any 4 numbers")))

    unique_times = set()

    for perm in permutations(digits,4):
        hour = perm[0]*10 + perm[1]
        minute = perm[2]*10 + perm[3]

        if 0 <= hour <= 23 and 0 <= minute <= 59:
            unique_times.add(f"{hour}:{minute}")
        
    print("Valid times:", unique_times)
clock()
```
