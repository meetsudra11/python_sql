### Prime Number Check
Given a number n, check if it’s prime or not.

````python
n = 10
prime = []
not_prime = []


for i in range(2,n+1):
    is_prime = True
    for j in range(2,int(i**0.5)+1):
        if i%j==0:
            is_prime = False
            break
    
    if is_prime:
        prime.append(i)
    else:
        not_prime.append(i)
    
print(prime)
print(not_prime)
````

### Strong Number (Factorial Sum of Digits)

A number is strong if the sum of factorial of digits equals the number.
Example: 145 = 1! + 4! + 5! = 145 ✅

````python
n = 145
fact_sum = 0

while n>0:
    last_num = n%10
    fact = 1
    for i in range(1,last_num+1):
        fact = fact * i
    fact_sum = fact_sum + fact
    n = n//10

print(fact_sum)
````

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

`Given List:`
```
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

`Output:`
```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```
