# default arguements

```python
def power(a,b):
    return a**b

power(5,5)

# on passing only one value to it, the function will crash. 
# power(5)

# the solution is default parameters 

def power(a=1,b=1):
    return a**b

power() # this wont crash 
```

# Nested functions 

```python
def f():
    print("Inside f function")
    def gandu():
        print("Inside g function")
    gandu()

f()
```

```python
gandu()
```

```python
def g(x):
    def h():
        x = 'abc'
    x = x+1
    print('in g(x): x=',x)
    h()
    return x

x=3
z=g(x)
```

```python
def g(x):
    def h(x):
        x = x+1
        print("in h(x):x =",x)
    x = x+1
    print("in g(x):x=",x)
    h(x)
    return x

x = 3
z = g(x)
print("in main program scope:x=",x)
print("in main program scope:z=",z)
```
