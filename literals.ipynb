```python
a = 0b10101 # binary literals 
print(a)
b = 100 # decimal literals 
print(b)
c = 0o310 # octadecimal literals 
print(c)
d = 0x12c # hexadecimal literals 
print(d)
```

```python
# float literal 
float1 = 10.5 
float2 = 1.5e2
float3 = 1.5e-3
```

```python
print(float2)
print(float3)
```

```python
# complex literal 
x = 3.14j # u can directly initialise with imaginary number 
print(x.imag)
print(x.real)
```

```python
# string litereals 
string1 = 'meet'
string2 = "meet"
string3 = 'C'
multiline_str = """Hello my name \n is meet sudra """
```

```python
print(multiline_str)
```

```python
# raw string 
a = r"Hello my name is \n meet sudra"
print(a) # agar aapko html print karna hee brother toh aap ye kar sakta he use 
```

```python
# boolean literal 
a = True + 1  # implicit type conversion 
print(a)
b = False + 10 
print(b)
```

```python
# special literal 
a = None
print(a)
```
