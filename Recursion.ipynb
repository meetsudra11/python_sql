To create a function of product, we have two techniques. 
1. loops 
2. recursion 

```python
# product of given 2 numbers using loops 

def product(a,b):
    product = 0 
    for i in range(1,b+1): 
        product = product + a
    return product
```

```python
product(5,6)
```

```python
# product of given 2 numbers using recursion 

def product(a,b):
    if b == 1:
        return a
    else:
        return a + product(a,b-1)
    
print(product(3,4))
```

```python
# factorial of given number using loops  

def factorial(a):
    fact = 1
    for i in range(1,a+1):
        fact = fact * i
    return fact 

factorial(4)
```

```python
# factorial of given number using recursion

def factorial_rec(a):
    if a == 1:
        return 1
    else:
        return a * factorial_rec(a-1)
    
factorial_rec(4)
```

```python
# palindrome using loops 

def palindrome(a):
    if a == a[::-1]:
        return "text is palindrome"
    else:
        return "it is not palindrome"
```

```python
palindrome("helleh")
```

```python
def pal(a):
    if len(a) <= 1:
        return "Text is pal"
    else:
        if a[0]==a[-1]:
            return pal(a[1:-1])
        else:
            return "Text is not a pal"
```

```python
pal("abba") 
```
