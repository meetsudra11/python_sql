###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)] 
Output : [(5, 6, 7, 8), (6, 10), (7, 13)] 
```

````python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
unique = []

for i in test_list:
    unique.append(i[0])
unique = set(unique)

result = []
for i in unique:
    result.append([i])
    for j in test_list:
        if j[0] == i:
            result[-1].append(j[1])


print(list(map(tuple,result)))
````

###`Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
t1 = (1,2,3,0)
t2 = (1,1,2,3)

if len(t1) != len(t2):
    print("2 tuples are not same")
else:
    is_same = True
    for i in range(len(t1)):
        if t1[i] != t2[i]:
            is_same=False
            break
    if is_same:
        print("Tuples are same")
    else:
        print("Tuples are not same")
````

````python
from itertools import permutations
a = [1,2,3,5]

def clock():
    my_perm = list(set(permutations(a)))
    time = {}
    for i in my_perm:
        h1,h2,m1,m2 = i[0],i[1],i[2],i[3]
        hour = h1*10+h2
        minute = m1*10+m2

        if hour<24 and minute<60:
            time[hour] = minute
    return len(time)
````

R = "BAABA", 
V = [2, 4, 1, 1, 2]

A --> B (2) 
B_bal =  + 2
A_bal =  - 2 (needs 2)

B --> A (4)
B_bal =  + 2 - 4 (needs 2)
A_bal =  - 2 + 4

B --> A (1)
B_bal =  + 2 - 4 - 1 (needs 3)
A_bal =  - 2 + 4 + 1

A --> B (1)
B_bal =  + 2 - 4 - 1 + 1 (needs 2)
A_bal =  - 2 + 4 + 1 - 1

B -- A (2)
B_bal =  + 2 - 4 - 1 + 1 - 2 (needs 4)
A_bal =  - 2 + 4 + 1 - 1 + 2

````python
R = "BAABA"
V = [2,4,1,1,2]

def min_bal(p,a):
    A_bal = 0
    B_bal = 0

    A_min_bal = 0
    B_min_bal = 0 

    for rec,amount in zip(p,a):
        if rec == 'A':
            A_bal += amount
            B_bal -= amount
        else :
            B_bal += amount 
            A_bal -= amount 

        if A_bal<0:
            A_min_bal = max(A_min_bal,-A_bal)
        if B_bal<0:
            B_min_bal = max(B_min_bal,-B_bal)

    return [A_min_bal,B_min_bal]

min_bal(R,V)
````

````python
S = 'CBABD'
stack = []

def solution(S):
    for i in S:
        if stack:
            top = stack[-1]
            if (top=='A' and i=='B') or (top=='B' and i=='A'):
                stack.pop()
                continue
            if (top=='C' and i=='D') or (top=='D' and i=='C'):
                stack.pop()
                continue
        stack.append(i)
    return ''.join(stack)

solution(S)
````

````python
A = [-1, 3, -4, 5, 1, -6, 2, 1]
def equilibrium_index(A):
    index = []
    for i in range(0,len(A)):
        if sum(A[:i]) == sum(A[i+1:]):
            index.append(A[i])
        else:
            continue
    if len(index)==0:
        return -1
    return index

equilibrium_index(A)
````

````python
A = [-1, 3, -4, 5, 1, -6, 2, 1]

def eqilibrium(A):
    total_sum = sum(A)
    left_sum = 0

    for i in range(len(A)):
        right_sum = total_sum-left_sum-A[i]
        if right_sum == left_sum:
            return A[i]
        else:
            left_sum = left_sum + A[i]
    return -1

eqilibrium(A)
````

````python
import numpy as np
A = [-1, 3, -4, 5, 1, -6, 2, 1]

def solution(A):
    total_sum = sum(A)
    left_sum = 0
    my_list = []
    distances = []
    for i in range(len(A)):
        right_sum = total_sum-left_sum-A[i]
        if left_sum == right_sum:
            my_list.append(A[i])
        distances.append(np.abs(left_sum-right_sum))
        left_sum += A[i]
    if len(my_list) == 0:
        return -1
    return distances

solution(A) 

````
